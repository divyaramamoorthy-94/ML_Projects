```python
import pandas as pd
from scipy.stats import shapiro, levene, f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd

alpha = 0.05
ALLOY_COL = "Alloy"
DENTIST_COL = "Dentist"
RESPONSE_COL = "Response"

df = df[[ALLOY_COL, DENTIST_COL, RESPONSE_COL]].dropna().copy() ## Keeping only required columns and remove missing rows
df[ALLOY_COL] = df[ALLOY_COL].astype(str) ## Convert to string type
df[DENTIST_COL] = df[DENTIST_COL].astype(str) ## Convert to string type
df[RESPONSE_COL] = pd.to_numeric(df[RESPONSE_COL], errors="coerce") ## Convert to numeric type
df = df.dropna(subset=[RESPONSE_COL]) ## Droping rows where response is non-numeric/missing after conversion.

print("\nHYPOTHESES (for each alloy separately):") ## These hypotheses apply independently within each alloy subset.
print("H0: Mean hardness is the same for all dentists.")
print("Ha: At least one dentist has a different mean hardness.")

for alloy in sorted(df[ALLOY_COL].unique()):  ## Gets unique alloy values from the alloy column which ensures stable orderand loops through each alloy one at a time.
    print("\n" + "=" * 75)
    print(f"ALLOY: {alloy}")
    print("=" * 75)
    sub = df[df[ALLOY_COL] == alloy].copy() ##Creates subset sub containing only rows for current alloyand copy() avoids chained lateral warnings.
    groups = [] ## will store response arrays per dentist.
    dentists = [] ## dentist names/IDs for display.
    for d, g in sub.groupby(DENTIST_COL): ## Groups sub by dentist for each dentist,d - dentist label,g - DataFrame of that dentist’s rows.
        dentists.append(d) ## Saves dentist name/ID in list.
        groups.append(g[RESPONSE_COL].values) ## Extracts that dentist’s response values as a NumPy array, also appends the groups.
    print(f"Sample size: {len(sub)}") ## Prints sample size.
    print(f"Dentists: {', '.join(dentists)}") ## Prints dentist names/IDs.

    print("\n[Assumption Check] Normality (Shapiro-Wilk by dentist):") ## Checks normality for each dentist.
    for d, g in sub.groupby(DENTIST_COL): ## Groups sub by dentist for each dentist,d - dentist label,g - DataFrame of that dentist’s rows.
        w, p_norm = shapiro(g[RESPONSE_COL].values) ## Performs Shapiro-Wilk normality test on response values.
        flag = "OK" if p_norm > alpha else "Possible non-normality"
        print(f"  Dentist {d}: W={w:.4f}, p={p_norm:.4f} -> {flag}") ## Prints W-statistic, p-value, and flag.
    lev_stat, lev_p = levene(*groups, center="mean") ## Performs Levene test on groups,unpacks list into separate group arguments and uses mean-centered Levene version which returns Levene statistic and lev_p: p-value for equal-variance assumption.
    lev_flag = "Equal variances plausible" if lev_p > alpha else "Variances may differ" ## Interprets Levene result:p > 0.05: equal variances plausible, p <= 0.05: variances may be unequal.
    print(f"\n[Assumption Check] Levene test: stat={lev_stat:.4f}, p={lev_p:.4f} -> {lev_flag}") ## Prints Levene test results.

    f_stat, p_anova = f_oneway(*groups) ## Performs one-way ANOVA which runs one-way ANOVA comparing dentist mean, F statistic, and p-value for overall mean difference
    print(f"\n[ANOVA] F-statistic={f_stat:.4f}, p-value={p_anova:.4f}") ## Prints ANOVA results.
    if p_anova < alpha: ## Checks if p-value is less than alpha.
        print("Decision: Reject H0")
        print("Conclusion: Implant hardness differs among dentists for this alloy.")
    else:
        print("Decision: Fail to reject H0")
        print("Conclusion: No statistically significant hardness difference among dentists for this alloy.")
    print("\n[Tukey HSD] Pairwise dentist comparison:") ## Performs Tukey HSD post-hoc test.
    tukey = pairwise_tukeyhsd( ## Performs Tukey HSD post-hoc test.
        endog=sub[RESPONSE_COL], ## Response values.
        groups=sub[DENTIST_COL], ## Dentist labels.
        alpha=alpha ## Significance level.
    )
    print(tukey)
    significant_pairs = int(tukey.reject.sum())
    if significant_pairs == 0:
        print("Result: No significant dentist pairs.")
    else:
        print(f"Result: {significant_pairs} dentist pair(s) show significant difference.")
```

```python
import pandas as pd
from scipy.stats import shapiro, levene, f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd

ALLOY_COL = "Alloy"
METHOD_COL = "Method"
RESPONSE_COL = "Response"


df = df[[ALLOY_COL, METHOD_COL, RESPONSE_COL]].dropna().copy() ## Keep needed columns and clean
df[ALLOY_COL] = df[ALLOY_COL].astype(str) ## Convert to string type
df[METHOD_COL] = df[METHOD_COL].astype(str) ## Convert to string type
df[RESPONSE_COL] = pd.to_numeric(df[RESPONSE_COL], errors="coerce")
df = df.dropna(subset=[RESPONSE_COL])

alpha = 0.05

print("Hypotheses for each alloy:") ## Hypotheses apply independently within each alloy subset.
print("H0: Mean hardness is the same across all methods.")
print("Ha: At least one method has a different mean hardness.")

for alloy in sorted(df[ALLOY_COL].unique()): ## Gets unique alloy values from the alloy column which ensures stable order and loops through each alloy one at a time.
    print("\n" + "=" * 75)
    print(f"ALLOY: {alloy}")
    print("=" * 75)

    sub = df[df[ALLOY_COL] == alloy].copy() ## Creates subset sub containing only rows for current alloy and copy() avoids chained lateral warnings.

    grouped = list(sub.groupby(METHOD_COL))
    method_names = [m for m, _ in grouped]
    response_groups = [g[RESPONSE_COL].values for _, g in grouped]
    print(f"Methods: {', '.join(method_names)}")
    print(f"Total observations: {len(sub)}")

    print("\n[Assumption] Normality check (Shapiro-Wilk by method):")
    for method, g in grouped: ## Groups sub by method for each method,method - method label,g - DataFrame of that method’s rows.
        w, p_norm = shapiro(g[RESPONSE_COL].values)
        status = "OK" if p_norm > alpha else "Possible non-normality"
        print(f"  Method {method}: W={w:.4f}, p={p_norm:.4f} -> {status}")
    lev_stat, lev_p = levene(*response_groups, center="mean")
    lev_status = "Equal variances plausible" if lev_p > alpha else "Variances may differ"
    print(f"\n[Assumption] Levene test: stat={lev_stat:.4f}, p={lev_p:.4f} -> {lev_status}")

    f_stat, p_anova = f_oneway(*response_groups)
    print(f"\n[ANOVA] F-statistic={f_stat:.4f}, p-value={p_anova:.4f}")

    if p_anova < alpha:
        print("Decision: Reject H0")
        print("Conclusion: Hardness differs across methods for this alloy.")

        print("\n[Tukey HSD] Pairwise method comparison:")
        tukey = pairwise_tukeyhsd(
            endog=sub[RESPONSE_COL],
            groups=sub[METHOD_COL],
            alpha=alpha
        )
        print(tukey)

        sig_pairs = int(tukey.reject.sum())
        if sig_pairs == 0:
            print("Result: No significant method pairs (despite significant ANOVA).")
        else:
            print(f"Result: {sig_pairs} method pair(s) differ significantly.")
    else:
        print("Decision: Fail to reject H0")
        print("Conclusion: No statistically significant hardness difference across methods for this alloy.")
        print("Pairwise identification not required (ANOVA not significant).")
```

4.4 How does the hardness of implants vary depending on dentists and methods together?

```python
import pandas as pd  # for data handling
from scipy.stats import shapiro, levene  # for assumption checks
import statsmodels.api as sm  # for ANOVA table
from statsmodels.formula.api import ols  # for fitting ANOVA model
from statsmodels.stats.multicomp import pairwise_tukeyhsd  # for post-hoc pairwise tests

ALLOY_COL = "Alloy"  # alloy column name
DENTIST_COL = "Dentist"  # dentist column name
METHOD_COL = "Method"  # method column name
RESPONSE_COL = "Response"  # response/hardness column name
alpha = 0.05  # significance level

df = df[[ALLOY_COL, DENTIST_COL, METHOD_COL, RESPONSE_COL]].dropna().copy()  # keep needed columns and drop missing rows
df[ALLOY_COL] = df[ALLOY_COL].astype(str)  # keep Alloy as categorical string
df[DENTIST_COL] = df[DENTIST_COL].astype(str)  # keep Dentist as categorical string
df[METHOD_COL] = df[METHOD_COL].astype(str)  # keep Method as categorical string
df[RESPONSE_COL] = pd.to_numeric(df[RESPONSE_COL], errors="coerce")  # convert Response to numeric
df = df.dropna(subset=[RESPONSE_COL])  # drop rows where Response could not be converted

print("H0(1): Mean hardness is same across dentists.")  # null for Dentist main effect
print("H0(2): Mean hardness is same across methods.")  # null for Method main effect
print("H0(3): No Dentist x Method interaction effect.")  # null for interaction
print("Ha: At least one of the above null hypotheses is false.")  # alternative statement

for alloy in sorted(df[ALLOY_COL].unique()):  # loop alloy-wise
    print("\n" + "=" * 80)
    print(f"Alloy: {alloy}")
    print("=" * 80)

    sub = df[df[ALLOY_COL] == alloy].copy()  # subset for current alloy only
    sub["Dentist_Method"] = sub[DENTIST_COL] + "_" + sub[METHOD_COL]  # interaction-level group labels

    print("\n[Assumption] Normality (Shapiro-Wilk) for each Dentist x Method group:")
    grouped_dm = sub.groupby([DENTIST_COL, METHOD_COL])  # group by interaction cells
    for name, g in grouped_dm:
        vals = g[RESPONSE_COL].values  # current group response values
        if len(vals) >= 3:  # Shapiro needs at least 3 values
            w, p_norm = shapiro(vals)  # normality test for this cell
            flag = "OK" if p_norm > alpha else "Possible non-normality"  # decision label
            print(f"Group {name}: W={w:.4f}, p={p_norm:.4f} -> {flag}")  # print result
        else:
            print(f"Group {name}: n={len(vals)} (too small for Shapiro)")  # small-sample note

    response_groups = [g[RESPONSE_COL].values for _, g in grouped_dm]  # list of all cell responses
    if len(response_groups) >= 2:  # Levene requires at least 2 groups
        lev_stat, lev_p = levene(*response_groups, center="mean")  # homogeneity of variance test
        lev_flag = "Equal variances plausible" if lev_p > alpha else "Variances may differ"  # decision label
        print(f"\n[Assumption] Levene test: stat={lev_stat:.4f}, p={lev_p:.4f} -> {lev_flag}")  # print result
    else:
        lev_p = float("nan")  # placeholder if test cannot run
        print("\n[Assumption] Levene test not possible (insufficient groups).")  # informative message

    model = ols(f"{RESPONSE_COL} ~ C({DENTIST_COL}) * C({METHOD_COL})", data=sub).fit()  # fit two-way model
    anova_table = sm.stats.anova_lm(model, typ=2)  # Type-II ANOVA table
    print("\n[Two-way ANOVA Results]")
    print(anova_table)  # show full ANOVA table (sum_sq, df, F, p-value)

    p_dentist = anova_table.loc[f"C({DENTIST_COL})", "PR(>F)"]  # p-value for Dentist main effect
    p_method = anova_table.loc[f"C({METHOD_COL})", "PR(>F)"]  # p-value for Method main effect
    p_inter = anova_table.loc[f"C({DENTIST_COL}):C({METHOD_COL})", "PR(>F)"]  # p-value for interaction effect

    print("\n[Decisions @ alpha = 0.05]")
    print(f"Dentist effect: {'Significant' if p_dentist < alpha else 'Not significant'} (p={p_dentist:.4f})")  # dentist decision
    print(f"Method effect: {'Significant' if p_method < alpha else 'Not significant'} (p={p_method:.4f})")  # method decision
    print(f"Interaction effect: {'Significant' if p_inter < alpha else 'Not significant'} (p={p_inter:.4f})")  # interaction decision

    if p_dentist < alpha:  # run Tukey only if dentist main effect is significant
        print("\n[Tukey HSD] Dentist pairs:")
        tukey_dentist = pairwise_tukeyhsd(endog=sub[RESPONSE_COL], groups=sub[DENTIST_COL], alpha=alpha)  # dentist pairwise
        print(tukey_dentist)  # print dentist comparisons
    else:
        print("\n[Tukey HSD] Dentist pairs skipped (Dentist effect not significant).")  # skip message

    if p_method < alpha:  # run Tukey only if method main effect is significant
        print("\n[Tukey HSD] Method pairs:")
        tukey_method = pairwise_tukeyhsd(endog=sub[RESPONSE_COL], groups=sub[METHOD_COL], alpha=alpha)  # method pairwise
        print(tukey_method)  # print method comparisons
    else:
        print("\n[Tukey HSD] Method pairs skipped (Method effect not significant).")

    if p_inter < alpha:  # run interaction-level Tukey only if interaction is significant
        print("\n[Tukey HSD] Interaction-level pairs (Dentist_Method):")
        tukey_inter = pairwise_tukeyhsd(endog=sub[RESPONSE_COL], groups=sub["Dentist_Method"], alpha=alpha)  # combo pairwise
        print(tukey_inter)  # print interaction-level comparisons
    else:
        print("\n[Tukey HSD] Interaction-level pairs skipped (Interaction effect not significant).")
```
